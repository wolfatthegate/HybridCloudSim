```python
import pandas as pd
import numpy as np
import random
```

```python
def json_to_csv(input_json, output_csv): 
    df = pd.read_json(input_json)

    if 'rotations' in df.columns:
        df['rotations'] = df['rotations'].apply(lambda x: ', '.join(x) if isinstance(x, list) else x)

    df.to_csv(output_csv, index=False)
    print(f"Successfully converted {input_json}, to {output_csv}")
```

```python
input_json = '5000-batch1.json'
# input_json2 = '5000-batch2.json'

df = pd.read_json(input_json)
# df2 = pd.read_json(input_json2)

# df = pd.concat([df, df2], ignore_index=True)

df.rename(columns={'qubits': 'num_qubits'}, inplace=True)
df.rename(columns={'shots': 'num_shots'}, inplace=True)
df['req_iterations'] = np.random.randint(4, 10, size=len(df))
df['cpu_units'] = np.random.randint(4,8, size=len(df))
df['mem_bw'] = np.random.randint(4,8, size=len(df))
df['priority'] = 2
df = df[df['depth'] < 100]
df.insert(0, 'job_id', range(1, len(df) + 1))
df['arrival_time'] = np.cumsum([random.expovariate(0.35) for _ in range(len(df))])
```
